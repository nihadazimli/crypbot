```python
from datetime import datetime, timedelta

# Function to fetch daily data from API for the past 90 days
def fetch_daily_data(symbol):
    # Calculate the start and end time
    end_time = datetime.combine(datetime.now().date(), datetime.min.time())  # Yesterday midnight
    start_time = end_time - timedelta(days=90)  # 90 days ago

    # Convert timestamps to milliseconds
    start_time_ms = int(start_time.timestamp() * 1000)
    end_time_ms = int(end_time.timestamp() * 1000)

    # Fetch data from API
    api_response = session.get_kline(
        category="spot",
        symbol=symbol,
        interval="D",  # Daily interval
        start=start_time_ms,
        end=end_time_ms
    )
    candle_list = api_response['result']['list']

    # Define columns
    columns = ['startTime', 'openPrice', 'highPrice', 'lowPrice', 'closePrice', 'volume', 'turnover']
    
    # Create a DataFrame
    df = pd.DataFrame(candle_list, columns=columns)

    # Convert startTime to datetime and set it as index
    df['startTime'] = pd.to_datetime(df['startTime'], unit='ms')
    
    return df
```

### Joining the seperate coin data in master_df

```python
import glob

directory = '/Users/teleixixie/Documents/20_40_clean/data/1d_data/'  

# Define the pattern to match files containing "_20230401_20240401" in their names
pattern = os.path.join(directory, '*_last_90d*')

# Use glob to get all files matching the pattern
matching_files = glob.glob(pattern)
```

```python
master_df_90d = pd.DataFrame()

for file_path in matching_files:
    
    # Find the file and print the name 
    
    file_name = os.path.basename(file_path)
    print(f'Found file: {file_name}')
    
    # Load it to df and calculate RSI 
    
    main_df = pd.read_csv('data/1d_data/' + file_name)
    main_df = main_df[['startTime','closePrice','lowPrice','volume']]
    if len(main_df) > 0:
        coin_name = file_name.split('_')[0]
        main_df['startTime'] = pd.to_datetime(main_df['startTime']).dt.round('15T')

        main_df.sort_values(by = 'startTime', ascending = True, inplace = True)
        main_df.reset_index(drop = True, inplace = True)

        main_df['coin_name'] = coin_name
        main_df['vol_usdt'] = main_df['volume']*main_df['closePrice']
        main_df['median_vol_usdt'] = main_df['vol_usdt'].median()

    # only include the coins the have on average above 1M median volume
    
    if main_df['median_vol_usdt'].median() > 1000000:
        master_df_90d = pd.concat([main_df, master_df_90d])
```

```python
coin_list = master_df_90d[['coin_name','median_vol_usdt']].drop_duplicates()

coin_list['median_vol_usdt_in_m'] = coin_list['median_vol_usdt']/1000000
```

```python
coin_list[['coin_name','median_vol_usdt_in_m']].sort_values(by = 'median_vol_usdt_in_m', ascending = False).to_csv('coins_list.csv')
```
